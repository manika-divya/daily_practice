```python
# print 1 to n numbers

n = int(input('enter the number:'))
for i in range(1,n+1):
    print(i,end = " ")
```

```python
# print n to 1 numbers

n = int(input('enter the number:'))
for i in range(n,0,-1):
    print(i,end = " ")
```

```python
# sum of n numbers

sum = 0
n = int(input('enter the number:'))
for i in range(n+1):
    sum += i
print(f"sum of n numbers is {sum}")
```

```python
# multiplication table

n = int(input('enter the number:'))
for i in range(1,10+1):
    print(f"{n} * {i} = {n*i}")
```

```python
# count digits in a number

n = int(input('enter the number:'))
count = 0
n = str(n)
for i in n:
    count += 1
print(count)
```

```python
# reverse number

n = int(input('enter the number:'))
s = str(n)
print(s[::-1])
```

```python
# sum of digits in a number

n = int(input('enter the number:'))
sum = 0
n1 = str(n)
for i in n1:
    i = int(i)
    sum += i
print(sum)
```

```python
# product of digits in a number

n = int(input('enter the number:'))
prod = 1
s = str(n)
for i in s:
    i = int(i)
    prod *= i
print(prod)
```

```python
# print first and last digits

n = int(input('enter the number:'))
s = str(n)
print(f"first number {s[0]} , last number {s[-1]}")
```

```python
# palindrome number

n = int(input('enter the number:'))
s = str(n)
s1 = s[::-1]

if s == s1:
    print('palindrome')
else:
    print('not a palindrome')
```

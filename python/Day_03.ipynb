```python
# Prime number

n = int(input('enter the number:'))

if n<=1:
    print('not a prime number')
else:
    for i in range(2,int(n**0.5)+1):
        if n%i == 0:
            print('not a prime number')
            break
        else:
            print('it is a prime number')
            break
```

```python
# Armstrong number
n = int(input('enter the number:'))
temp = n
digits = len(str(n))
total = 0

while temp>0:
    digit = temp%10
    total += digit**digits
    temp //= 10

if total == n:
    print('it is a armstrong number')
else:
    print('it is not a armstrong number')
```

```python
# Perfect Number
n = int(input('enter the number:'))
sum_div = 0

for i in range(1,n):
    if n%i == 0:
        sum_div +=i
if sum_div == n:
    print('it is a perfect number')
else:
    print('it is not a perfect number')
    
```

```python
# Strong number
import math
n = int(input('enter the number:'))
temp =n
fact =0

while temp>0:
    digit = temp % 10
    fact += math.factorial(digit)
    temp //= 10
if fact == n:
    print('it is a strong number')
else:
    print('it is not a strong number')
```

```python
# Automorphic number
n = int(input('enter the number:'))
square = n*n

if str(square).endswith(str(n)):
    print('Automorphic')
else:
    print('not automorphic')
```

```python
# Harshad number
# 18 -> 1+8 = 9 -> 18%9 ==0

n = int(input('enter the number:'))
digits_sum = sum(int(i) for i in str(n))

if n%digits_sum ==0:
    print('harshad')
else:
    print('not a harshad')
```

```python
# Neon number
# 9*9 =81 -> 8+1 = 9

n = int(input('enter the number:'))

square = n*n
digits_sum = sum(int(i) for i in str(square))

if digits_sum == n:
    print('neon')
else:
    print('not a neon')
```
